```python
#1. Bernoulli Trial – Zomato Login

#A Bernoulli trial has only two possible outcomes:

#Success = 1
#Failure = 0

#Probability of successful login = 0.85

import numpy as np

p = 0.85

results = np.random.binomial(1, p, 20)

print("Login Results:")
print(results)
```

```python
#2. Binomial Distribution – UPI Payments

#There are 5 UPI payment attempts.

#Number of attempts = 5
#Number of successful payments = 3
#Probability of success = 0.9

#Using the binomial distribution:

from scipy.stats import binom

n = 5
k = 3
p = 0.9

probability = binom.pmf(k, n, p)

print("Probability of exactly 3 successful payments:", probability)
print("Percentage:", probability * 100, "%")
```

```python
#3. Poisson Distribution – Spotify Playlists

#Average number of playlists created per minute:

#λ = 4

#We need to find the probability of exactly 6 playlists:

from scipy.stats import poisson

lambda_value = 4
k = 6

probability = poisson.pmf(k, lambda_value)

print("Probability of exactly 6 playlists:", probability)
print("Percentage:", probability * 100, "%")
```

```python
#4. Poisson Distribution – Flipkart Product Reviews

#Flipkart receives an average of 10 reviews per hour.

#We can simulate the number of reviews received during 12 randomly selected hours.

import numpy as np
import matplotlib.pyplot as plt

lambda_value = 10
hours = 12

reviews = np.random.poisson(lambda_value, hours)

print("Reviews received in 12 hours:")
print(reviews)

plt.plot(range(1, hours + 1), reviews, marker='o')

plt.xlabel("Hour")
plt.ylabel("Number of Reviews")
plt.title("Flipkart Product Reviews per Hour")

plt.show()
```

5. Real-World Analytics Use Case – BookMyShow

AI-generated use case:

BookMyShow can use the Poisson distribution to model the number of movie ticket bookings received by a particular cinema in one hour. For example, if a cinema receives an average of 20 bookings per hour, the Poisson distribution can estimate the probability of receiving exactly 25 bookings in the next hour.

Why Poisson distribution fits:

Poisson distribution is suitable because we are counting the number of events occurring within a fixed time period. Here, ticket bookings are individual events and the average booking rate can be used to estimate different booking counts.
