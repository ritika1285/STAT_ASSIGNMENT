```python
#Task 1

#Question:Given a list of delivery times (in minutes) for 20 Zomato orders, calculate the 25th, 50th, and 75th percentiles using Python's numpy.percentile() function.
#Answer:

import numpy as np

delivery_times = [
    25, 32, 28, 45, 30,
    35, 40, 27, 31, 29,
    50, 38, 33, 26, 42,
    36, 30, 34, 28, 39
]

p25 = np.percentile(delivery_times, 25)
p50 = np.percentile(delivery_times, 50)
p75 = np.percentile(delivery_times, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

```python
#Task 2

#Question:Create a Python script that takes an array of IPL player scores and divides them into quartiles using numpy.quantile(). Print the values for Q1, Q2 (median), and Q3.
#Answer:

import numpy as np

scores = [
    12, 45, 67, 23, 89,
    34, 56, 78, 10, 91,
    43, 65, 32, 54, 76
]

Q1 = np.quantile(scores, 0.25)
Q2 = np.quantile(scores, 0.50)
Q3 = np.quantile(scores, 0.75)

print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
```

```python
#Task 3

#Question:Given a list of daily step counts from your fitness app for one month, calculate the Interquartile Range (IQR) in Python and explain what it tells you about your activity consistency.
#Answer:

import numpy as np

steps = [
    6500, 7200, 8000, 6000, 9000,
    7500, 8200, 6800, 7100, 7600,
    8300, 6200, 7000, 7900, 8500,
    7300, 6700, 8100, 7400, 7800,
    6900, 8600, 7700, 6400, 7200,
    8000, 7500, 8300, 7100, 7600
]

Q1 = np.quantile(steps, 0.25)
Q3 = np.quantile(steps, 0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
#Task 4

#Question:Write a Python function to detect outliers in a list of Swiggy order amounts using the IQR method. Print the outlier values.
#Answer:

import numpy as np

def detect_outliers(order_amounts):
    Q1 = np.percentile(order_amounts, 25)
    Q3 = np.percentile(order_amounts, 75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    outliers = [
        value for value in order_amounts
        if value < lower_limit or value > upper_limit
    ]

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outliers)


order_amounts = [
    250, 300, 280, 350, 400,
    320, 290, 450, 310, 275,
    330, 360, 295, 310, 5000
]

detect_outliers(order_amounts)
```

```python
#Task 5

#Question:Using matplotlib, create a boxplot of your weekly screen time (in hours) for the past 8 weeks. Label the axes and highlight any outliers.
#Answer:

import matplotlib.pyplot as plt

screen_time = [24, 26, 25, 28, 27, 30, 26, 45]

plt.boxplot(
    screen_time,
    patch_artist=True,
    flierprops=dict(
        marker='o',
        markerfacecolor='red',
        markersize=8
    )
)

plt.xlabel("Screen Time")
plt.ylabel("Hours per Week")
plt.title("Weekly Screen Time for 8 Weeks")

plt.show()
```
