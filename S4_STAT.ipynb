```python
#Task 1 — Range of Daily Step Counts

steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200]

range_steps = max(steps) - min(steps)

print("Daily step counts:", steps)
print("Range of steps:", range_steps)
```

```python
#Task 2 — Variance and Standard Deviation of Food Orders

import statistics

orders = [250, 400, 350, 300, 500]

variance = statistics.variance(orders)
standard_deviation = statistics.stdev(orders)

print("Order amounts:", orders)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

```python
#Task 3 — Coefficient of Variation of Mobile Data Usage

import statistics

data_usage = [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]

mean = statistics.mean(data_usage)
standard_deviation = statistics.stdev(data_usage)

cv = (standard_deviation / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
print("Coefficient of Variation:", cv, "%")
```

```python
#Task 4 — Standard Deviation of Instagram Likes

import statistics

likes = [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]

mean = statistics.mean(likes)
standard_deviation = statistics.stdev(likes)

print("Mean likes:", mean)
print("Standard deviation:", standard_deviation)
```
