```python
#Task 1 — Calculate Mean of Daily Step Counts

#Question: Create an array of 7 daily step counts and calculate the mean.

# Daily step counts
steps = [5200, 6800, 7500, 6100, 8200, 7000, 6500]

# Calculate mean
mean_steps = sum(steps) / len(steps)

print("Daily steps:", steps)
print("Mean steps:", mean_steps)
```

```python
#Task 2 — Median Delivery Time

import statistics

delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

median_time = statistics.median(delivery_times)

print("Delivery times:", delivery_times)
print("Median delivery time:", median_time, "minutes")
```

```python
#Task 3 — Mode of YouTube Video Genres

from collections import Counter

genres = [
    'Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog',
    'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy'
]

genre_count = Counter(genres)

mode_genre = genre_count.most_common(1)[0][0]

print("Genre counts:", genre_count)
print("Most watched genre:", mode_genre)
```

```python
#Task 4 — Mean vs Median of UPI Transactions

import statistics

transactions = [100, 120, 105, 110, 5000, 115, 108]

mean_amount = statistics.mean(transactions)
median_amount = statistics.median(transactions)

print("Transactions:", transactions)
print("Mean:", mean_amount)
print("Median:", median_amount)
```
