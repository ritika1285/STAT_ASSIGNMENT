```python
#Task 1: Calculate Pearson correlation between danceability and popularity of songs.

#Answer:

import pandas as pd

# Create dataset
data = {
    "song": [
        "Song A", "Song B", "Song C", "Song D", "Song E",
        "Song F", "Song G", "Song H", "Song I", "Song J",
        "Song K", "Song L"
    ],
    "danceability": [0.80, 0.75, 0.60, 0.90, 0.55, 0.70,
                     0.85, 0.50, 0.65, 0.95, 0.72, 0.58],
    "energy": [0.75, 0.80, 0.60, 0.90, 0.55, 0.70,
               0.85, 0.50, 0.65, 0.95, 0.78, 0.62],
    "popularity": [85, 80, 65, 95, 55, 75,
                   90, 50, 70, 98, 78, 60]
}

df = pd.DataFrame(data)

# Calculate Pearson correlation
correlation = df["danceability"].corr(df["popularity"], method="pearson")

print("Pearson Correlation:", correlation)
```

Interpretation:

A positive correlation means that songs with higher danceability tend to have higher popularity in this sample. Correlation does not prove that danceability causes popularity.

```python
#Task 2: Calculate Spearman correlation between IPL players' runs and Instagram followers.

#Answer:

import pandas as pd
from scipy.stats import spearmanr

# Create IPL player dataset
data = {
    "player": ["Player A", "Player B", "Player C", "Player D",
               "Player E", "Player F", "Player G", "Player H",
               "Player I", "Player J"],

    "total_runs": [700, 500, 900, 300, 800, 400, 1000, 600, 200, 750],

    "instagram_followers": [800000, 500000, 1500000, 200000,
                            1200000, 350000, 2000000, 650000,
                            100000, 1000000]
}

df_ipl = pd.DataFrame(data)

# Calculate Spearman correlation
corr, p_value = spearmanr(
    df_ipl["total_runs"],
    df_ipl["instagram_followers"]
)

print("Spearman Correlation:", corr)
print("P-value:", p_value)
```

Interpretation:

A positive Spearman correlation indicates that players with higher runs generally have more Instagram followers in this sample.

A correlation close to +1 indicates a strong positive relationship.

This relationship does not prove that scoring more runs directly increases followers.

```python
#Task 3: Create a correlation matrix for Flipkart product features.

#Answer:

import pandas as pd

# Create product dataset
data = {
    "price": [500, 1000, 1500, 2000, 2500,
              3000, 3500, 4000, 4500, 5000],

    "user_rating": [3.5, 3.8, 4.0, 4.2, 4.5,
                    4.1, 4.3, 4.6, 4.4, 4.8],

    "number_of_reviews": [50, 100, 150, 200, 300,
                          250, 400, 500, 450, 600]
}

df_products = pd.DataFrame(data)

# Create correlation matrix
corr_matrix = df_products.corr()

print("Correlation Matrix:")
print(corr_matrix)
```

```python
#Task 4: Visualize the correlation matrix using a heatmap.

#Answer:

import matplotlib.pyplot as plt
import seaborn as sns

# Create heatmap
plt.figure(figsize=(8, 5))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    cbar=True,
    linewidths=0.5
)

plt.title("Flipkart Product Features Correlation Heatmap")
plt.show()
```

Task 5: List two potential pitfalls when interpreting correlation coefficients.

Answer:

1. Correlation does not mean causation

Correlation between two variables does not prove that one variable causes the other.

Example: On YouTube, videos with more views may also have more likes. However, more views do not necessarily cause more likes. Both may be influenced by video quality, promotion or audience interest.

2. Outliers can affect correlation

An outlier is an unusually high or low value that can change the correlation coefficient.

Example: On Zomato, most restaurants may have ratings between 3.5 and 4.5, but one restaurant may have an unusually high number of reviews and a low rating. This unusual record can affect the correlation between ratings and review counts.

Conclusion: Correlation helps identify relationships between variables, but we should consider other factors, outliers and the actual context before drawing conclusions.
