```python
from scipy.stats import ttest_1samp, ttest_ind, ttest_rel
```

```python
#Task 1. One-sample t-test

#Question: Using Python and the scipy.stats library, perform a one-sample t-test to check if the average delivery time for Zomato orders in your city is significantly different from 30 min
#Answer:

import numpy as np
from scipy.stats import ttest_1samp

# Sample delivery times
delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]

# Hypothesized average delivery time
population_mean = 30

# Perform one-sample t-test
t_stat, p_value = ttest_1samp(delivery_times, population_mean)

print("Average Delivery Time:", np.mean(delivery_times))
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
#Task 2. Independent two-sample t-test

#Question: Simulate two independent samples of daily step counts: one group before installing a fitness app and one group after installation. Use Python to perform an independent t-test and determine if average steps increased.

#Answer:

import numpy as np
from scipy.stats import ttest_ind

# Daily steps before installing the fitness app
before = [4000, 4500, 5000, 4200, 4800,
          3900, 4600, 4300, 4100, 4700]

# Daily steps after installing the fitness app
after = [5500, 6000, 5800, 6200, 5700,
         6500, 5900, 6100, 6300, 5600]

# Perform independent t-test
t_stat, p_value = ttest_ind(after, before)

print("Average Steps Before:", np.mean(before))
print("Average Steps After:", np.mean(after))
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant increase in daily steps")
else:
    print("No significant increase in daily steps")
```

```python
#Task 3. Paired t-test

#Question: Simulate paired data for 10 users showing their average screen time before and after enabling Focus Mode. Conduct a paired t-test to check whether Focus Mode significantly reduced screen time.

#Answer:

import numpy as np
from scipy.stats import ttest_rel

# Screen time before Focus Mode (hours)
before = [6.5, 7.0, 5.5, 8.0, 6.0,
          7.5, 6.8, 5.8, 7.2, 6.4]

# Screen time after Focus Mode (hours)
after = [5.0, 5.5, 4.5, 6.5, 5.0,
         6.0, 5.5, 4.8, 5.8, 5.2]

# Perform paired t-test
t_stat, p_value = ttest_rel(before, after)

print("Average Screen Time Before:", np.mean(before))
print("Average Screen Time After:", np.mean(after))
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Focus Mode significantly reduced screen time")
else:
    print("No significant reduction in screen time")
```

```python
#Task 4. Compare Zomato and Swiggy ratings

#Question: Compare the average ratings of two food delivery apps (Zomato vs Swiggy) based on 15 randomly selected user ratings for each app. Write Python code to perform the appropriate t-test and explain why you chose that test.

#Answer:

import numpy as np
from scipy.stats import ttest_ind

# Ratings of 15 Zomato users
zomato = [4.2, 4.5, 4.0, 3.8, 4.3,
          4.6, 4.1, 4.4, 3.9, 4.5,
          4.2, 4.0, 4.3, 4.6, 4.1]

# Ratings of 15 Swiggy users
swiggy = [3.8, 4.0, 3.9, 4.2, 3.7,
          4.1, 3.8, 4.0, 3.9, 4.1,
          3.7, 4.2, 3.9, 4.0, 3.8]

# Perform independent t-test
t_stat, p_value = ttest_ind(zomato, swiggy)

print("Average Zomato Rating:", np.mean(zomato))
print("Average Swiggy Rating:", np.mean(swiggy))
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference in ratings")
else:
    print("No significant difference in ratings")
```

Significant difference in ratings

Why did we choose an independent t-test?

We chose an independent t-test because Zomato and Swiggy ratings come from two separate groups of users. We want to compare their average ratings to check whether a statistically significant difference exists.

```python
#Task 5. One-sample t-test using PhonePe data

#Question: Use ChatGPT to generate Python code that performs a one-sample t-test on a dataset of your choice, such as average daily UPI transactions per user on PhonePe. Run the code and paste both the code and output.

#Answer:

import numpy as np
from scipy.stats import ttest_1samp

# Daily UPI transactions per user
transactions = [4, 5, 6, 5, 7,
                3, 6, 5, 4, 8,
                5, 6, 4, 7, 5]

# Hypothesized population mean
population_mean = 5

# Perform one-sample t-test
t_stat, p_value = ttest_1samp(transactions, population_mean)

print("Average Daily Transactions:", np.mean(transactions))
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

Interpretation:

The average daily UPI transactions per user are approximately 5.33. Since the p-value (0.4170) is greater than 0.05, we fail to reject the null hypothesis. There is not enough statistical evidence to conclude that the average daily transactions are different from 5.
