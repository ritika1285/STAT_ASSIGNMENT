```python
#1. Normal Distribution – Sales Amounts

#Generate 1000 random sales amounts with a mean of 500 and standard deviation of 100.

import numpy as np
import matplotlib.pyplot as plt

mean = 500
std = 100

sales = np.random.normal(mean, std, 1000)

plt.hist(sales, bins=30, density=True)

plt.xlabel("Sales Amount")
plt.ylabel("Density")
plt.title("Normal Distribution of Sales Amounts")

plt.show()
```

```python
#2. Z-Score of 9500 Steps

#X = 9500
#Mean (μ) = 8000
#Standard deviation (σ) = 1200

#Formula:

#Z = (X - μ) / σ

#Calculation:

#Z = (9500 - 8000) / 1200

#Z = 1500 / 1200

#Z = 1.25

steps = 9500
mean = 8000
std = 1200

z_score = (steps - mean) / std

print("Z-score:", z_score)
```

```python
#3. Standard Normal Curve

#A standard normal distribution has:

Mean = 0
#Standard deviation = 1
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.plot(x, y)

x_fill = np.linspace(-1, 1, 500)
y_fill = norm.pdf(x_fill, 0, 1)

plt.fill_between(x_fill, y_fill)

plt.xlabel("Z-score")
plt.ylabel("Density")
plt.title("Standard Normal Distribution")

plt.show()
```

```python
#4. Probability of Delivery Taking Less Than 25 Minutes

#Mean = 30 minutes
#Standard deviation = 5 minutes
#X = 25 minutes

#Using scipy.stats.norm.cdf():

from scipy.stats import norm

mean = 30
std = 5
x = 25

probability = norm.cdf(x, mean, std)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

Explain in simple words why the normal distribution is important for analyzing real-world app data like sales, delivery times, or user ratings.
