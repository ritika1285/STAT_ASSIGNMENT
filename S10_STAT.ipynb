```python
#1. Random Sampling of 20 Users
import random

user_ids = list(range(1, 201))

sample = random.sample(user_ids, 20)

print("Random Sample of 20 Users:")
print(sample)
```

```python
#2. Stratified Sampling of Flipkart Orders

#orders are divided into groups based on city. The sample is then selected proportionally from each city.

import pandas as pd
import numpy as np

np.random.seed(42)

cities = np.random.choice(
    ["Ahmedabad", "Surat", "Vadodara"],
    size=1000,
    p=[0.50, 0.30, 0.20]
)

orders = pd.DataFrame({
    "Order_ID": range(1, 1001),
    "City": cities
})

sample_size = 60

city_counts = orders["City"].value_counts()
sample = pd.DataFrame()

for city, count in city_counts.items():
    city_sample_size = round((count / len(orders)) * sample_size)

    city_data = orders[orders["City"] == city]

    selected = city_data.sample(
        n=city_sample_size,
        random_state=42
    )

    sample = pd.concat([sample, selected])

print("Stratified Sample:")
print(sample)

print("\nSample Count by City:")
print(sample["City"].value_counts())
```

```python
#3. Systematic Sampling of Zomato Reviews

#Systematic sampling selects every 10th review.

def systematic_sampling(reviews, interval=10):
    sample = reviews[::interval]
    return sample


reviews = [f"Review_{i}" for i in range(1, 501)]

sample = systematic_sampling(reviews, 10)

print("Number of Reviews:", len(sample))
print("Sample Reviews:")
print(sample)
```

4. Effect of Sample Size on Accuracy

Suppose there are 10,000 Swiggy delivery records and we want to estimate the average delivery time.

Different sample sizes can be compared as follows:

Sample Size	Percentage of Population	Expected Accuracy
50	0.5%	Lower
200	2%	Better
500	5%	Higher

For example, suppose the actual average delivery time of all 10,000 orders is 30 minutes.

A sample of 50 orders might give:

Sample mean = 32 minutes

A sample of 200 orders might give:

Sample mean = 30.8 minutes

A sample of 500 orders might give:

Sample mean = 30.3 minutes

As the sample size increases, the sample is generally more likely to represent the population accurately, so the estimated average tends to become more stable.

Conclusion: A sample of 500 records would generally provide a more reliable estimate than 50 records, although it requires more time and resources to analyze.

```python
#5. Stratified Sampling of IPL Fans

import pandas as pd
import numpy as np

np.random.seed(42)

teams = [
    "MI", "CSK", "RCB",
    "KKR", "GT", "RR"
]

fans = pd.DataFrame({
    "Fan_ID": range(1, 201),
    "Favorite_Team": np.random.choice(
        teams,
        200,
        p=[0.25, 0.20, 0.20, 0.15, 0.10, 0.10]
    )
})

sample_size = 40
stratified_sample = []

for team in teams:

    team_fans = fans[fans["Favorite_Team"] == team]

    proportion = len(team_fans) / len(fans)

    required = round(proportion * sample_size)

    selected_fans = team_fans.sample(
        n=required,
        random_state=42
    )

    stratified_sample.append(selected_fans)

sample = pd.concat(stratified_sample)

print("Stratified Sample:")
print(sample)

print("\nFans Selected from Each Team:")
print(sample["Favorite_Team"].value_counts())
```
