SESSION 11 – Central Limit Theorem (CLT)

The Central Limit Theorem (CLT) states that when we take a large number of random samples from a population and calculate the mean of each sample, the distribution of those sample means becomes approximately normal (bell-shaped), even if the original population is not normally distributed.

Importance of CLT :

Helps understand large datasets using smaller samples.

Makes it easier to estimate population averages.

Helps in statistical analysis and decision-making.

Supports confidence intervals and hypothesis testing.

```python
#Task 1. Simulate CLT using Python

#Question: Simulate the Central Limit Theorem in Python by taking 1000 random samples of size 30 from a skewed dataset (number of likes on Instagram posts). Calculate the mean of each sample and plot a histogram.

#Answer:

import numpy as np
import matplotlib.pyplot as plt

# Create a skewed dataset (Instagram likes)
np.random.seed(42)
likes = np.random.exponential(scale=50, size=10000)

# Take 1000 random samples of size 30
sample_means = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    sample_means.append(np.mean(sample))

# Plot histogram of sample means
plt.hist(sample_means, bins=30, edgecolor='black')

plt.xlabel("Sample Mean of Instagram Likes")
plt.ylabel("Frequency")
plt.title("Central Limit Theorem (Sample Size = 30)")
plt.show()
```

#Task 2. Explain why the sampling distribution becomes normal

Question: Explain in your own words why the sampling distribution of the mean becomes approximately normal, even if the original data is not, using an example from an app you use daily.

Answer:

The Central Limit Theorem explains that when we take many random samples from a dataset and calculate their averages, these averages tend to form a normal distribution.

For example, consider Zomato order amounts. Some customers may order food worth ₹100, while others may order ₹1,000 or more. This makes the original order amount data uneven or right-skewed.

If we repeatedly select 30 orders and calculate the average amount for each sample, the averages will form a more balanced, bell-shaped distribution.

This happens because individual high and low values tend to balance out when we calculate the average of many observations.

```python
#Task 4. Compare sample size 5 and sample size 30

#Question: Modify the CLT simulation code to use a smaller sample size (n=5) and compare the histogram of sample means with n=30.

#Answer:

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create skewed Instagram likes dataset
likes = np.random.exponential(scale=50, size=10000)

# Sample size 5
means_5 = []

for i in range(1000):
    sample = np.random.choice(likes, size=5)
    means_5.append(np.mean(sample))

# Sample size 30
means_30 = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    means_30.append(np.mean(sample))

# Plot both histograms
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(means_5, bins=30, edgecolor='black')
plt.title("Sample Size = 5")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(means_30, bins=30, edgecolor='black')
plt.title("Sample Size = 30")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

Observation:

Sample size 5: The histogram is more spread out and may still look right-skewed.

Sample size 30: The histogram looks more bell-shaped and is more concentrated around the average.

As sample size increases, the variation between sample means decreases and the distribution becomes more approximately normal.

```python
#Task 5. Demonstrate CLT using an exponential distribution

#Question: Use ChatGPT to generate Python code that demonstrates the Central Limit Theorem using random samples from a non-normal distribution, such as exponential. Run the code, paste the output plot, and briefly describe what you observe.

#Answer:

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate exponential distribution
data = np.random.exponential(scale=2, size=10000)

# Calculate means of 1000 random samples
sample_means = []

for i in range(1000):
    sample = np.random.choice(data, size=30)
    sample_means.append(np.mean(sample))

# Plot original data and sample means
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(data, bins=40, edgecolor='black')
plt.title("Original Exponential Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(sample_means, bins=30, edgecolor='black')
plt.title("Sampling Distribution of Means")
plt.xlabel("Sample Means")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
